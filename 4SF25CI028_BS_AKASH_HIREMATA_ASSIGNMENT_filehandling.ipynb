```python
file1 = open("document.txt", "w")
file1.write("Python is easy to learn and Python is powerful")

file1 = open("document.txt", "r")

text = file1.read()

word_count = len(text.split())

print("Number of words:", word_count)
```

```python
old_word = "hello"
new_word = "hi"

file2 = open("data.txt", "w")
file2.write("hello everyone hello friends hello students")

file2 = open("data.txt", "r")

text = file2.read()

text = text.replace(old_word, new_word)

file2 = open("data.txt", "w")

file2.write(text)

print("All occurrences of", old_word, "replaced with", new_word)
```

```python
file3 = open("reverse.txt", "w")
file3.write("Apple\nBanana\nMango")

file3 = open("reverse.txt", "r")

lines = file3.readlines()

for line in lines[::-1]:
    print(line.strip())
```

```python
file4 = open("stats.txt", "w")
file4.write("Python is easy\nPython is fun")

file4 = open("stats.txt", "r")

text = file4.read()

line_count = len(text.splitlines())
word_count = len(text.split())
char_count = len(text)

print("Lines:", line_count)
print("Words:", word_count)
print("Characters:", char_count)
```
