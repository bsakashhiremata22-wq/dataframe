```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=["A", "B", "C"])
df1["Product"] = df1["A"] * df1["B"]
df1
```

```python
df = pd.DataFrame({
    "A": np.random.randint(1,10,4),
    "B": np.random.randint(1,10,4),
    "C": np.random.randint(1,10,4)
})

df["Row_Sum"] = df.sum(axis=1)

display(df)
```

```python
df = pd.DataFrame({
    "A": np.random.randint(1,10,5),
    "B": np.random.randint(1,10,5),
    "C": np.random.randint(1,10,5)
})

df.loc[1,"A"] = np.nan
df.loc[3,"B"] = np.nan

df = df.fillna(df.mean())

df
```

```python
df = pd.DataFrame({
    "A": np.random.randint(1,10,6),
    "B": np.random.randint(1,10,6),
    "C": np.random.randint(1,10,6),
    "D": np.random.randint(1,10,6)
})

df.loc[1,"A"] = np.nan
df.loc[4,"C"] = np.nan

df = df.dropna()

df
```

```python
df = pd.DataFrame({
    "Category": np.random.choice(["A","B","C"],10),
    "Value": np.random.randint(1,100,10)
})

result = df.groupby("Category").agg(
    Sum=("Value","sum"),
    Mean=("Value","mean")
)

result
```

```python
df = pd.DataFrame({
    "Product":["P1","P2","P3","P4","P5","P6"],
    "Category":["A","B","A","C","B","C"],
    "Sales":np.random.randint(100,1000,6)
})

result = df.groupby("Category")[["Sales"]].sum()

result
```

```python
df1 = pd.DataFrame({
    "ID":[1,2,3,4],
    "Name":["Ram","Sam","Tom","John"]
})

df2 = pd.DataFrame({
    "ID":[1,2,3,4],
    "Marks":[85,90,88,92]
})

result = pd.merge(df1,df2,on="ID")

result
```

```python
index = pd.MultiIndex.from_tuples([
    ("A","X"),
    ("A","Y"),
    ("B","X"),
    ("B","Y")
])

df = pd.DataFrame({
    "Value":np.random.randint(1,100,4)
},index=index)

result = df.groupby(level=[0,1]).sum()

result
```

```python
df = pd.DataFrame({
    "A": np.random.randint(1,10,5),
    "B": np.random.randint(1,10,5),
    "C": np.random.randint(1,10,5)
})

df = df * 2

df
```

```python
df = pd.DataFrame({
    "A": np.random.randint(1,10,6),
    "B": np.random.randint(1,10,6),
    "C": np.random.randint(1,10,6)
})

df["Total"] = df.apply(
    lambda x: x["A"] + x["B"] + x["C"],
    axis=1
)

df
```

```python
s = pd.Series([
    "python",
    "pandas",
    "numpy",
    "data",
    "science"
])

df = pd.DataFrame({
    "Original": s,
    "Uppercase": s.str.upper()
})

df
```
